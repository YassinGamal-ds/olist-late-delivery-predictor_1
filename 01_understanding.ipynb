# 01 — Data Understanding

**Project:** Olist Late-Delivery Predictor

In this notebook, we inspect the original Olist tables before combining them. The aim is to understand what each table represents, its size, column types, missing values, and the keys used to connect tables.

**Dataset:** [Brazilian E-Commerce Public Dataset by Olist](https://www.kaggle.com/datasets/olistbr/brazilian-ecommerce)

Download and extract the CSV files into `data/raw/` before running this notebook.

```python
from pathlib import Path
import pandas as pd
import numpy as np

ROOT = Path.cwd()
if not (ROOT / "data").exists() and (ROOT.parent / "data").exists():
    ROOT = ROOT.parent
RAW = ROOT / "data" / "raw"

files = {
    "orders": "olist_orders_dataset.csv",
    "items": "olist_order_items_dataset.csv",
    "customers": "olist_customers_dataset.csv",
    "sellers": "olist_sellers_dataset.csv",
    "products": "olist_products_dataset.csv",
    "geolocation": "olist_geolocation_dataset.csv",
    "payments": "olist_order_payments_dataset.csv",
    "reviews": "olist_order_reviews_dataset.csv",
}

tables = {}
for name, filename in files.items():
    path = RAW / filename
    if path.exists():
        tables[name] = pd.read_csv(path)
        print(f"{name:15} {tables[name].shape}")
    else:
        print(f"{name:15} NOT FOUND: {path}")
```

## 1. Preview the tables

```python
for name, df in tables.items():
    print(f"\\n===== {name.upper()} =====")
    display(df.head(3))
```

## 2. Basic structure and data types

```python
for name, df in tables.items():
    print(f"\\n===== {name.upper()} =====")
    print("Rows and columns:", df.shape)
    display(df.dtypes.astype(str).to_frame("data_type"))
```

## 3. Missing values

```python
missing_summary = []
for name, df in tables.items():
    missing = df.isna().sum()
    for column, count in missing.items():
        missing_summary.append({
            "table": name,
            "column": column,
            "missing_count": int(count),
            "missing_percent": round(count / len(df) * 100, 2) if len(df) else 0
        })
missing_df = pd.DataFrame(missing_summary)
display(missing_df.query("missing_count > 0").sort_values(
    ["table", "missing_percent"], ascending=[True, False]
))
```

## 4. Primary keys and relationships

```python
key_checks = {
    "orders": ["order_id"],
    "items": ["order_id", "order_item_id"],
    "customers": ["customer_id"],
    "sellers": ["seller_id"],
    "products": ["product_id"],
    "payments": ["order_id"],
    "reviews": ["review_id"],
}
for table, cols in key_checks.items():
    if table in tables:
        existing = [c for c in cols if c in tables[table].columns]
        print(table, {c: tables[table][c].is_unique for c in existing})
```

## 5. Initial observations

Use the outputs above to write what you actually notice.
